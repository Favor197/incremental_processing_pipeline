```python
import pandas as pd
```

```python
input_path ="/Volumes/workspace/default/datasets/Online Retail.xlsx"
output_path ="/Volumes/incremental_sales/default/data/archive"
dbutils.fs.mkdirs(output_path)
```

```python
df= pd.read_excel(input_path)
df.head()
```

```python
unique_year_month = set(zip(df['InvoiceDate'].dt.year, df['InvoiceDate'].dt.month))
```

```python
for ind in unique_year_month:
    year = ind[0]
    month = ind[1]
    data = df[(df['InvoiceDate'].dt.year == year) & (df['InvoiceDate'].dt.month == month)]
    name = f"sales_{year}_{month:02d}"
    file_path = f"{output_path}/{name}.csv"
    data.to_csv(file_path, index=False)
    print(f"File {name} saved to {file_path}")
    
print("File partitioning complete")
```
