```python
input_filepath ='/Volumes/incremental_sales/default/data/archive/'
output_filepath ='/Volumes/incremental_sales/default/data/landing/'
```

```python
files = [f for f in dbutils.fs.ls(input_filepath) if f.name.endswith(".csv")]

if files:
    files.sort(key=lambda f: f.name)
    file_to_move = files[0]
    filepath = output_filepath+file_to_move.name
    dbutils.fs.mv(file_to_move.path, filepath)
    print(f"{file_to_move.name} moved to {filepath}")
else:
    print("No files found")
```
